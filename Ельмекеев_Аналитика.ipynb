```python
import pandas as pd

url = "https://docs.google.com/spreadsheets/d/1e3VQ9BR9ShL7E4nGTM2tpW2zmHIfGuxXYcYlkC7YZwU/export?format=csv"
df = pd.read_csv(url)
df.head(5)
```

```python
# Подсчитать количество каждого значения (True/False) в колонке 'is_cancel'
value_counts = df['is_cancel'].value_counts()

print(value_counts)
```

```python
# Фильтруем и оставляем только строки, где is_cancel == False
df_is_not_cancel = df[df['is_cancel'] == False]

df_is_not_cancel.shape
```

```python
# Заменил запятую на точку и потом перевел в числовой тип данных
df['total_price1']=df['total_price'].str.replace(',','.').astype(float)
```

```python
df['updated_at'] = pd.to_datetime(df['updated_at'], format='%d.%m.%Y %H:%M')
df= df[df['updated_at'] <= '2025-01-01']
df['year_month'] = df['updated_at'].dt.to_period('M')
df
```

```python
# Группировка по артикулу
df_grouped = df.groupby(['nm_id', 'year_month'], as_index=False).agg(
    total_sales=('total_price1', 'sum'),  # Сумма продаж по артикулу
    count_orders=('total_price1', 'size')  # Количество заказов (строчек) по артикулу
)

# Сортировка по сумме продаж от большого к меньшему
df_grouped = df_grouped.sort_values(by='year_month', ascending=False)

df_grouped
```

```python
df_december_2024 = df_grouped[df_grouped['year_month'] == '2024-12']
df_december_2024[['nm_id', 'total_sales', 'count_orders']].sort_values(by='nm_id')
```

```python
df_november_2024 = df_grouped[df_grouped['year_month'] == '2024-11']
df_november_2024[['nm_id', 'total_sales', 'count_orders']].sort_values(by='nm_id')
```

```python
df_october_2024 = df_grouped[df_grouped['year_month'] == '2024-10']
df_october_2024[['nm_id', 'total_sales', 'count_orders']].sort_values(by='nm_id')
```

```python
# Метод взвешенных средних (Exponentially Weighted Moving Average)
# Этот метод придаёт больший вес более недавним данным, что позволяет лучше учесть изменения трендов.

# Вычисляем EWMA для total_sales и count_orders для каждого товара
df_grouped['ewma_sales'] = df_grouped.groupby('nm_id')['total_sales'].transform(lambda x: x.ewm(span=3, adjust=False).mean())
df_grouped['ewma_orders'] = df_grouped.groupby('nm_id')['count_orders'].transform(lambda x: x.ewm(span=3, adjust=False).mean())
```

```python
df_grouped
```

```python
# Теперь можно сгруппировать по товарам и рассчитать план продаж для января 2025 года
# Прогноз на январь 2025 = последние значения EWMA для каждого товара
df_forecast = df_grouped.groupby('nm_id').head(1)[['nm_id', 'ewma_sales', 'ewma_orders']]

# Переименуем столбцы для финального вывода
df_forecast = df_forecast.rename(columns={'ewma_sales': 'forecast_sales_jan_2025', 'ewma_orders': 'forecast_orders_jan_2025'})

# Округляем значения в столбцах до целых чисел
df_forecast[['forecast_sales_jan_2025', 'forecast_orders_jan_2025']] = \
    df_forecast[['forecast_sales_jan_2025', 'forecast_orders_jan_2025']].round(0).astype(int)

# Выводим итоговый результат
df_forecast[['nm_id', 'forecast_sales_jan_2025', 'forecast_orders_jan_2025']].sort_values(by='nm_id')
```
